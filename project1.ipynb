```python
avg_total_revenue=df['Total Revenue'].mean()
avg_total_revenue
```

```python
grouped_mean=df.groupby('Product Category')['Total Revenue'].mean()
grouped_mean
```

```python
category_sales=df.groupby('Product Category')['Units Sold'].sum()
category_sales.plot(kind='bar')
plt.title("product category vs units sold")
plt.xlabel('product category')
plt.ylabel('units sold')
```

```python
profuct_name_revenue=df.groupby('Product Name')['Total Revenue'].sum()
profuct_name_revenue.plot(kind='line')
plt.title("product name vs total revenue")
plt.xlabel('product name')
plt.ylabel('total revenue')
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(data=df, x='Total Revenue', y='Units Sold')
plt.title('Total Revenue vs Units Sold')
plt.xlabel('Total Revenue')
plt.ylabel('Units Sold')
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

group_data = df.groupby(['Region', 'Payment Method'])['Total Revenue'].sum()

heatmap_data = group_data.unstack()

heatmap_data = heatmap_data.fillna(0)

sns.heatmap(heatmap_data)

plt.title("Region vs Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Region")

plt.show()
```
